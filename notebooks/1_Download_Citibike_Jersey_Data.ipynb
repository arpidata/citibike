## Create a Period Iterator

Define a helper function that generates a list of year-month periods (YYYYMM). These periods are later used to automate the download of multiple Citi Bike datasets.

```python
def period_iterator(year: list, start_m: int, stop_m: int) -> list:
    """
    Generate a list of year-month periods in YYYYMM format.
    """

    YEAR = year
    MONTH = [str(i + 1) if i + 1 > 9 else "0" + str(i + 1)
             for i in range(start_m, stop_m)]

    periods = []

    for i in YEAR:
        for j in MONTH:
            periods.append(i + j)

    return periods
```

## Generate Download Periods

Generate a list of monthly reporting periods that will be used to automate the download of Citi Bike Jersey City datasets.

```python
periods = period_iterator(
    year=["2025"],
    start_m=1,
    stop_m=12
)

periods
```

## Download and Extract Monthly Citi Bike Datasets

Loop through the selected reporting periods, download each monthly Citi Bike dataset, extract its contents, and remove the ZIP archive after extraction.

The code also handles both available ZIP file naming conventions used by Citi Bike.

```python
from pathlib import Path
from urllib.request import urlretrieve
from zipfile import ZipFile
from urllib.error import HTTPError, URLError

CITIBIKE_INDEX_URL = "https://s3.amazonaws.com/tripdata"

OUTPUT_DIR = "../data/citibike"
output_dir = Path(OUTPUT_DIR)
output_dir.mkdir(exist_ok=True)

for i in periods:

    try:
        file_name = f"JC-{i}-citibike-tripdata.csv.zip"
        url = f"{CITIBIKE_INDEX_URL}/{file_name}"

        zip_path = output_dir / file_name
        urlretrieve(url, zip_path)

    except (HTTPError, URLError, FileNotFoundError):
        file_name = f"JC-{i}-citibike-tripdata.zip"
        url = f"{CITIBIKE_INDEX_URL}/{file_name}"

        zip_path = output_dir / file_name
        urlretrieve(url, zip_path)

    with ZipFile(zip_path, "r") as zip_ref:
        zip_ref.extractall(output_dir)

    print(f"{file_name} Extracted")

    zip_path.unlink()

    print(f"{file_name} removed.")
```

## Load Monthly Citi Bike Datasets

Read all extracted CSV files from the project directory and load them into individual Pandas DataFrames. This step also allows us to inspect the column structure of each dataset before combining them.

```python
import glob
import numpy as np
import pandas as pd

file_names = glob.glob(f"{output_dir}/*.csv")

dfs = []
cols = []

for file_name in file_names:
    df = pd.read_csv(file_name)

    print(df.columns, 2 * "||", len(df.columns))

    cols.append(list(df.columns))
    dfs.append(df)
```

```python
citibike_df = pd.concat(dfs)

citibike_df.head()
```

## Save Combined Citi Bike Dataset

After combining all monthly Citi Bike files, we save the complete dataset
as a single CSV file for further analysis and data enrichment steps.

```python
citibike_df.to_csv("../data/citibike/JC2025.csv", index = False)
```
