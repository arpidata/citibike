```python
import pandas as pd

citibike_df = pd.read_csv(
    "../data/processed/JC2025_Enriched.csv"
)

citibike_df.head()
```

## Monthly Ride Analysis

We calculate the total number of Citi Bike rides for each month of 2025.

This summary helps identify seasonal trends and monthly changes in bike usage before creating visualizations.

```python
monthly_rides = (
    citibike_df
    .groupby("month", as_index=False)
    .agg(
        number_of_rides=("ride_id", "count")
    )
)

monthly_rides
```

## Visualize Monthly Ride Volume

We create a bar chart to compare the total number of Citi Bike rides
across different months in 2025.

This visualization highlights seasonal changes in bike usage and
helps identify the busiest and quietest months.

```python
fig = px.bar(
    monthly_rides,
    x="month",
    y="number_of_rides",
    title="Number of Citi Bike Rides per Month"
)

fig.update_layout(
    xaxis_title="Month",
    yaxis_title="Number of Rides",
)

fig.show()
```

## Seasonal Ride Analysis

We calculate the total number of Citi Bike rides for each season.

To ensure the results are displayed in chronological order, we define
the season order as Winter, Spring, Summer, and Autumn.

```python
season_rides = (
    citibike_df
    .groupby("season", as_index=False)
    .agg(
        number_of_rides=("ride_id", "count")
    )
)

season_order = ["Winter", "Spring", "Summer", "Autumn"]

season_rides["season"] = pd.Categorical(
    season_rides["season"],
    categories=season_order,
    ordered=True
)

season_rides = season_rides.sort_values("season")

season_rides
```

## Visualize Seasonal Ride Volume

We create a bar chart to compare the total number of Citi Bike rides
across the four seasons.

This visualization helps identify how bike usage changes throughout the year.

```python
fig = px.bar(
    season_rides,
    x="season",
    y="number_of_rides",
    title="Number of Citi Bike Rides per Season",
    text_auto=True
)

fig.update_layout(
    xaxis_title="Season",
    yaxis_title="Number of Rides"
)

fig.show()
```

## Top 10 Departure Stations

We identify the ten most frequently used departure stations based on
the total number of rides.

This analysis helps reveal the busiest Citi Bike stations and provides
insight into the most popular trip origins.

```python
top_start_stations = (
    citibike_df
    .dropna(subset=["start_station_name"])
    .groupby("start_station_name", as_index=False)
    .agg(
        number_of_departures=("ride_id", "count")
    )
    .sort_values("number_of_departures", ascending=False)
    .head(10)
)

top_start_stations
```

## Visualize Top 10 Departure Stations

We create a horizontal bar chart showing the ten most frequently used
Citi Bike departure stations.

The chart ranks stations by the total number of rides that started
from each location, helping identify the busiest bike stations in
Jersey City.

```python
fig = px.bar(
    top_start_stations.sort_values("number_of_departures"),
    x="number_of_departures",
    y="start_station_name",
    orientation="h",
    title="Top 10 Start Stations by Number of Departures",
    text_auto=True
)

fig.update_layout(
    xaxis_title="Number of Departures",
    yaxis_title="Start Station"
)

fig.show()
```

## Top 10 Arrival Stations

We identify the ten most frequently used destination stations based on
the total number of ride arrivals.

This analysis highlights the most popular destinations in the Citi Bike
network and complements the departure station analysis.

```python
top_end_stations = (
    citibike_df
    .dropna(subset=["end_station_name"])
    .groupby("end_station_name", as_index=False)
    .agg(
        number_of_arrivals=("ride_id", "count")
    )
    .sort_values("number_of_arrivals", ascending=False)
    .head(10)
)

top_end_stations
```

## Visualize Top 10 Arrival Stations

We create a horizontal bar chart showing the ten most frequently used
Citi Bike destination stations.

This visualization helps identify the busiest arrival locations and
understand where riders most commonly finish their trips.

```python
fig = px.bar(
    top_end_stations.sort_values("number_of_arrivals"),
    x="number_of_arrivals",
    y="end_station_name",
    orientation="h",
    title="Top 10 End Stations by Number of Arrivals",
    text_auto=True
)

fig.update_layout(
    xaxis_title="Number of Arrivals",
    yaxis_title="End Station"
)

fig.show()
```

## Daily Rides and Temperature Trend

We visualize the daily changes in Citi Bike usage and average
temperature over time using a dual-axis line chart.

This allows us to compare seasonal patterns and observe how changes
in temperature correspond with variations in daily ride volume.

```python
import plotly.graph_objects as go


fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=bike_weather_daily["date"],
        y=bike_weather_daily["number_of_rides"],
        mode="lines",
        name="Daily Rides",
        yaxis="y1"
    )
)

fig.add_trace(
    go.Scatter(
        x=bike_weather_daily["date"],
        y=bike_weather_daily["temperature_2m_mean"],
        mode="lines",
        name="Daily Average Temperature",
        yaxis="y2"
    )
)

fig.update_layout(
    title="Daily Rides and Daily Average Temperature",
    xaxis=dict(
        title="Day"
    ),
    yaxis=dict(
        title="Daily Rides",
        side="left"
    ),
    yaxis2=dict(
        title="Daily Average Temperature",
        overlaying="y",
        side="right"
    ),
    hovermode="x unified",
    legend=dict(
        orientation="h",
        yanchor="bottom",
        y=1.02,
        xanchor="right",
        x=1
    )
)

fig.show()
```
